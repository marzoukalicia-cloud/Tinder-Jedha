# PART 1: EXTRACTION AND DESCRIPTION

```python
# 0. IMPORTER LES LIBRAIRIES

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
# 1. CHARGEMENT ET DESCRIPTION DU DATASET
# ----------------------------------------------------------------

df = pd.read_csv("Speed+Dating+Data.csv", encoding="latin1")
df.head()

#df.info() # il y a 195 colonnes et 8 378 lignes
#df.dtypes # pour voir les types des colonnes
```

```python
# 2. STATISTIQUES DESCRIPTIVES DU DATASET 
# ---------------------------------------------------------------- 

df.describe(include="all")
#df["order"].describe(include="all")
#df.isna().sum() 
# attention: il y a beaucoups de valeurs manquantes dans certaines colonnes, 
# il faudra les traiter avant de faire des analyses.
```

```python
# 3. NETTOYAGE DES DONNEES
# ---------------------------------------------------------------- 

# Nettoyage des données manquantes et abérantes pour les variables d'intérets:
# match, dec_o, gender, age, samerace
# attr1_1, sinc1_1, intel1_1, fun1_1, amb1_1, shar1_1
# attr2_1, sinc2_1, intel2_1, fun2_1, amb2_1, shar2_1
# attr4_1, sinc4_1, intel4_1, fun4_1, amb4_1, shar4_1
# attr3_1, sinc3_1, intel3_1, fun3_1, amb3_1, shar3_1
# attr5_1, sinc5_1, intel5_1, fun5_1, amb5_1, shar5_1
# attr_o, sinc_o, intel_o, fun_o, amb_o, shar_o
# attr, sinc, intel, fun, amb, shar

# Normalisation des vagues 6 à 9
# Les vagues 6-9 utilisent une échelle 1-10 → on les convertit en 1-100

vagues_10 = [6, 7, 8, 9]
cols_to_fix = ['attr1_1', 'sinc1_1', 'intel1_1', 'fun1_1', 'amb1_1', 'shar1_1']

# On identifie uniquement les lignes concernées
mask = df['wave'].isin(vagues_10)

# On vérifie si les valeurs sont bien sur 1-10 (sécurité)
if df.loc[mask, cols_to_fix].max().max() <= 10:
    df.loc[mask, cols_to_fix] = df.loc[mask, cols_to_fix] * 10
    print("Normalisation effectuée pour les vagues 6 à 9.")
else:
    print("Les vagues 6 à 9 semblent déjà normalisées.")


# Nettoyage des revenus (String to Float)
if df['income'].dtype == 'object':
    df['income'] = df['income'].str.replace(',', '').astype(float)
df['income'] = df['income'].fillna(df['income'].median())
```

```python
# 4. Créer le genre pour le partner
df['gender_o'] = 1 - df['gender']

df["gender"] = df["gender"].replace({0: "Female", 1: "Male"})
df["gender_o"] = df["gender_o"].replace({0: "Female", 1: "Male"})
df.head()
```

```python
# 5. Création du dataset par individu (pour éviter les doublons)
profiles = df.drop_duplicates(subset='iid')
```

```python
# 6. Description de l'age par genre
profiles.groupby('gender')['age'].describe()

# boxplot de l'age par genre
fig=px.box(
    profiles, 
    x="gender", 
    y="age", 
    color="gender", 
    title="Distribution de l'âge par genre")
fig.show()

# histogramme de l'age par genre 
fig = px.histogram(
    profiles, 
    x="age", 
    color="gender",
    title="Distribution de l'âge par genre",
    labels={"age": "Âge", "gender_label": "Genre"},
    color_discrete_map={"Female": "red", "Male": "blue"},
    #opacity=0.7
)
fig.show()
```

Les femmes ont entre 19 et 55 ans avec une médiane de 26 ans et les hommes ont entre 18 et 42 ans avec une médiane de 27 ans.

```python
# 7. DESCRIPTION DES AUTRES VARIABLES ET SPEED DATING
# ---------------------------------------------------------------- 

# wave: entre 1 et 21
# round: number of people that met in wave -> entre 5 et 22
# position: station number where met partner -> entre 1 et 22
# order: the number of date that night when met partner -> entre 1 et 22

# VARIABLE CIBLE: 
# "match"
# 1 = la personne veut revoir l’autre,
# 0 = non

# VARIABLE DECISION POUR REVOIR L'AUTRE: 
# "dec_o"

# six attributes: Attractiveness, Sincerity, Intelligence, Fun, Ambition, and Shared Interests.

## TIME 1
# what do you want?
# attr1_1, sinc1_1, intel1_1, fun1_1, amb1_1, shar1_1

# What do you think the opposite sex looks for in a date?
# attr2_1, sinc2_1, intel2_1, fun2_1, amb2_1, shar2_1

# what you think MOST of your fellow men/women look for…?
# attr4_1, sinc4_1, intel4_1, fun4_1, amb4_1, shar4_1

# Self-assessment 
# attr3_1, sinc3_1, intel3_1, fun3_1, amb3_1, shar3_1

# How I think others see me
# attr5_1, sinc5_1, intel5_1, fun5_1, amb5_1, shar5_1

# partner
# attr_o, sinc_o, intel_o, fun_o, amb_o, shar_o
```

```python
# Description du taux de match
fig_f = px.pie(
    df["match"].value_counts(),
    names=df["match"].value_counts().index.map({0: "Pas de match", 1: "Match"}),
    values=df["match"].value_counts().values,
    title="Répartition des matchs")
fig_f.show()

# Calcul du taux de decision pour un second date par genre
dec_by_gender = df.groupby("gender")["dec"].value_counts().unstack().fillna(0)

# Camembert pour les femmes
fig_f = px.pie(
    dec_by_gender.loc["Female"],
    names=dec_by_gender.loc["Female"].index.map({0: "Pas de match", 1: "Match"}),
    values=dec_by_gender.loc["Female"].values,
    title="Répartition des décisions pour un second date chez les femmes"
)
fig_f.show()

# Camembert pour les hommes
fig_h = px.pie(
    dec_by_gender.loc["Male"],
    names=dec_by_gender.loc["Male"].index.map({0: "Pas de match", 1: "Match"}),
    values=dec_by_gender.loc["Male"].values,
    title="Répartition des décisions pour un second date chez les hommes"
)
fig_h.show()
```

Il y a donc 16.5% de match. Quand à la décision pour un second date, il y'a 36.5% de oui chez les femmes contre 47.4% chez les hommes.

### Q1: What are the least desirable attributes in a male partner? Does this differ for female partners?

```python
# Calculer la moyenne des attributs pour les partenaire hommes
attributes_o = ["attr_o", "sinc_o", "intel_o", "fun_o", "amb_o", "shar_o"]

mean_male_o = profiles[profiles["gender_o"] == "Male"][attributes_o].mean().reset_index()
mean_male_o.columns = ["attribute", "mean_male_score"]
mean_male_o = mean_male_o.sort_values(by="mean_male_score", ascending=True)
print(mean_male_o)
print("The least desirable attribute for a male partner is Shared Interest")
print("-------------------------------")

# Calculer la moyenne des attributs pour les partenaire femmes
mean_female_o = profiles[profiles["gender_o"] == "Female"][attributes_o].mean().reset_index()
mean_female_o.columns = ["attribute", "mean_female_score"]
mean_female_o = mean_female_o.sort_values(by="mean_female_score", ascending=True)
print(mean_female_o)
print("The least desirable attribute for a female partner is Shared Interest")
```

```python
# Utilisation de subplots pour comparer Hommes (1) et Femmes (0)
fig = make_subplots(rows=1, cols=2, subplot_titles=("Préférences : Hommes", "Préférences : Femmes"))

pref_cols = ['attr1_1', 'sinc1_1', 'intel1_1', 'fun1_1', 'amb1_1', 'shar1_1']
names = ['Attractivité', 'Sincérité', 'Intelligence', 'Humour', 'Ambition', 'Intérêts']

fig.add_trace(go.Bar(x=names, y=profiles[profiles.gender=="Male"][pref_cols].mean(), name="H"), row=1, col=1)
fig.add_trace(go.Bar(x=names, y=profiles[profiles.gender=="Female"][pref_cols].mean(), name="F"), row=1, col=2)

fig.update_layout(title_text="Attributs les moins désirables (Déclaration d'intention)", showlegend=False)
fig.show()
```

```python
# HEATMAP
# 1. On groupe par genre et on calcule la moyenne des attributs
heatmap_data = profiles.groupby('gender')[pref_cols].mean()

# 2. On renomme pour que ce soit lisible (Standard Plotly)
heatmap_data.columns = ['Beauté', 'Sincérité', 'Intelligence', 'Humour', 'Ambition', 'Intérêts']
heatmap_data.index = ['Femmes', 'Hommes']

# 3. Création de la Heatmap (Plus précise qu'une corrélation ici)
import plotly.express as px
fig = px.imshow(
    heatmap_data,
    text_auto=True, 
    aspect="auto",
    color_continuous_scale='RdPu',
    title="<b>Profil type des attentes :</b> Score moyen attribué par genre",
    labels=dict(x="Attributs", y="Genre", color="Importance (0-100)")
)
fig.show()
```

Conclusion Q1: L'analyse montre une différence entre les genres : alors que les hommes placent l'attractivité loin devant tous les autres critères, les femmes expriment des attentes plus équilibrées entre la sincérité, l'intelligence et l'humour. Pour les deux genres, l'ambition et les intérêts communs sont déclarés comme les moins prioritaires.

### Q2: How important do people think attractiveness is in potential mate selection vs. its real impact?

```python
# variables to analyze : attr1_1 vs attr_o
# variable of interest : match

#Q2_1
df_plot = df.copy()
df_plot["match_label"] = df_plot["match"].map({0: "Non", 1: "Oui"})

fig = px.box(
    df_plot,
    x="match_label",
    y="attr1_1",
    color="match_label",
    title="Importance déclarée de l’attractivité selon match"
)
fig.show()

fig2 = px.box(
    df_plot,
    x="match_label",
    y="attr_o",
    color="match_label",
    title="Attractivité réelle reçue selon match"
)
fig2.show()

#Q2_3
df[["attr1_1", "attr_o", "match"]].corr()
df[['attr_o', 'match']].corr()
```

En résumé : l’attractivité est clairement un facteur déterminant dans la probabilité de match.


### Q3: Are shared interests more important than a shared racial background?

```python
# for shared interest, variable: shar_o or shar1_1
# for same race, variable: samerace
```

```python
# Q3_1 — Taux de match selon même origine

# Calcul du taux de match
race_match = (
    df.groupby("samerace")["match"]
      .mean()
      .reset_index()
)

# Renommage pour plus de clarté
race_match["samerace"] = race_match["samerace"].map({
    0: "Origine différente",
    1: "Même origine"
})

print("Taux de match selon l'origine :")
print(race_match)

# Visualisation simple
fig = px.bar(
    race_match,
    x="samerace",
    y="match",
    title="Taux de match selon l'origine",
    text="match"
)

# Affichage en pourcentage
fig.update_traces(texttemplate="%{text:.1%}", textposition="outside")
fig.update_yaxes(title="Taux de match", tickformat=".0%")
fig.update_xaxes(title="Origine")

fig.show()
```

Le taux de match est presque identique entre :
- Origine différente : 16.1%
- Même origine : 17.1%

La différence est d’environ 1 point, ce qui indique que la race n’a quasiment aucun effet sur la probabilité de match.

```python
# label pour match
df["match_label"] = df["match"].map({0: "No match", 1: "Match"})

# Q3_2 Share interest
match_shar=df.groupby("match_label")[["shar1_1", "shar_o"]].mean()
print(match_shar)

# Q3_3
fig=px.box(df_plot, x="match_label", y="shar1_1", color="match_label", title="Shared interest importance by match")
fig.show()

fig2=px.box(df_plot, x="match_label", y="shar_o", color="match_label", title="Shared interest importance for the partner by match")
fig2.show()

#Q3_4
df[["shar1_1", "shar_o", "match"]].corr()
```

Les boxplots montrent que :
- les personnes qui matchent déclarent accorder plus d’importance aux intérêts communs (shar1_1)
- elles reçoivent aussi des scores plus élevés de la part de leurs partenaires (shar_o)

La corrélation confirme :
- shar_o ↔ match = 0.27 → effet modéré et réel
- shar1_1 ↔ match ≈ 0 → effet très faible

```python
# RACE VS INTÉRÊTS

# Comparons le taux de match si même race vs corrélation d'intérêts
race_impact = df.groupby('samerace')['match'].mean()

# Utilisation d'un Box plot pour voir la distribution des intérêts chez ceux qui matchent
fig_race = px.box(df, x='match', y='int_corr', color='samerace',
             title="Impact de la similarité raciale vs Intérêts communs",
             labels={'int_corr': 'Corrélation des intérêts', 'match': 'Match (0/1)'})
fig_race.show()
```

Conclusion Q3: Les résultats montrent que l’origine (même race ou non) a un effet très faible sur le taux de match : la différence entre les deux groupes est d’environ 1 point de pourcentage, ce qui est négligeable.
En revanche, les deux boxplots indiquent que les personnes qui ont matché ont tendance à accorder plus d’importance aux intérêts partagés, surtout lorsqu’il s’agit de l’importance qu’elles attribuent aux intérêts du partenaire.
Les intérêts partagés semblent jouer un rôle plus important que l’origine dans la probabilité de match.

### Q4: Can people accurately predict their own perceived value in the dating market?

```python
# six attributes: Attractiveness, Sincerity, Intelligence, Fun, Ambition, and Shared Interests.
# attr, sinc, intel, fun, amb, shar

# Ce que les gens pensent d’eux-mêmes
# attr1_1 : importance qu’ils accordent à l’attractivité
# attr : comment ils s’auto‑évaluent (selon les versions du dataset)
# Ce que les autres pensent d’eux
# attr_o : attractivité perçue par le partenaire
# match : résultat réel
```

```python
#Q4_1

print("Attractiveness: self vs partner rating by match:")
print(df.groupby("match")[["attr", "attr1_1", "attr_o"]].mean(), "\n")
print("------------------------------")
print("Sincerity: self vs partner rating by match:")
print(df.groupby("match")[["sinc", "sinc1_1", "sinc_o"]].mean(), "\n")
print("------------------------------")
print("Intelligence: self vs partner rating by match:")
print(df.groupby("match")[["intel", "intel1_1", "intel_o"]].mean(), "\n")
print("------------------------------")
print("Fun: self vs partner rating by match:")
print(df.groupby("match")[["fun", "fun1_1", "fun_o"]].mean(), "\n")
print("------------------------------")
print("Ambition: self vs partner rating by match:")
print(df.groupby("match")[["amb", "amb1_1", "amb_o"]].mean(), "\n")
print("------------------------------")
print("Shared interests: self vs partner rating by match:")
print(df.groupby("match")[["shar", "shar1_1", "shar_o"]].mean(), "\n")
```

- les personnes qui matchent ont des self‑ratings plus élevés
- les personnes qui ne matchent pas ont des self‑ratings plus bas   

Cela montre que les gens qui plaisent réellement ont tendance à avoir une meilleure perception d’eux‑mêmes.

```python
# Q4_2 

fig = px.scatter(
    df,
    x="attr1_1",
    y="attr_o",
    trendline="ols",
    title="Self-rated vs partner-rated attractiveness"
)
fig.show()
```

Le scatter plot montre :

- une droite de régression ascendante
- une relation faible à modérée mais réelle
👉 Les personnes qui se trouvent attirantes sont aussi jugées attirantes par les partenaires.

Conclusion Q4: Les participants semblent relativement précis dans l’évaluation de leur propre valeur sur le marché du dating.
Les self‑ratings sont fortement alignés avec les ratings donnés par les partenaires, comme le montrent les moyennes et le scatterplot.
Cependant, ils ont tendance à surestimer l’importance de leurs qualités (importance déclarée très élevée), ce qui montre une légère distorsion entre ce qu’ils pensent être déterminant et ce qui influence réellement le match.
👉 En résumé : les gens évaluent assez bien comment ils sont perçus, mais surestiment l’importance de leurs propres attributs dans le succès du dating.
